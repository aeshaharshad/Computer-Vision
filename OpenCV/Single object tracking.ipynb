```python

import cv2

# Check available modules
print(cv2.__version__)

TrDict = {}

try:
    TrDict['csrt'] = cv2.legacy.TrackerCSRT_create
    TrDict['kcf'] = cv2.legacy.TrackerKCF_create
    TrDict['boosting'] = cv2.legacy.TrackerBoosting_create
    TrDict['mil'] = cv2.legacy.TrackerMIL_create
    TrDict['tld'] = cv2.legacy.TrackerTLD_create
    TrDict['medianflow'] = cv2.legacy.TrackerMedianFlow_create
    TrDict['mosse'] = cv2.legacy.TrackerMOSSE_create
except AttributeError:
    print("Legacy tracker API not found. Install opencv-contrib-python.")
```

```python
tracker = TrDict ['csrt']()
```

```python
v=cv2.VideoCapture(r"D:\istockphoto-1426881162-640_adpp_is.mp4")
```

```python
ret,frame=v.read()
cv2.imshow('Frame',frame)
bb=cv2.selectROI('Frame',frame)
tracker.init(frame,bb)
```

```python
while True:
    ret,frame=v.read()
    if not ret:
        break
    (success,box)=tracker.update(frame)
    if success:
        (x,y,w,h)=[int (a) for a in box]
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
    cv2.imshow('Frame',frame)
    key=cv2.waitKey(5) & 0xFF
    if key == ord('q'):
        break
v.release()
cv2.destroyAllWindows
```
