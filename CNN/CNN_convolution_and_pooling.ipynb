# Low-level implementation of convolution and pooling without built in functions for understanding

```python
import numpy as np
```

```python
def f_padd(I,p): #Function for zero padding on image
    numRows=I.shape[0]
    numCols=I.shape[1]
    zeroRows=np.zeros((p,numCols))
    I=np.vstack((zeroRows,I))
    I=np.vstack((I,zeroRows))
    zeroCols = np.zeros((numRows+2*p,p))
    I=np.hstack((zeroCols,I))
    I=np.hstack((I,zeroCols))
    return I
```

```python
def f_conv2d(I,K,p): #returning Convolution
    fSize = K.shape[0]
    I2=f_padd(I,p)
    numRows=I2.shape[0]
    numCols=I2.shape[1]
    C = np.zeros((numRows-2*p,numCols-2*p))
    for i in range (numRows-fSize+1):
        for j in range (numCols-fSize+1):
            A = I2[i:i+fSize , j:j+fSize]
            C[i,j]=(A.flatten()*K.flatten()).sum()
    return C
    
```

```python
def f_Relu(C):
    C[C<0] =0
    return C
```

```python
def f_pool(C):
    r=C.shape[0]
    c=C.shape[1]
    S=np.zeros((int(r/2),int(c/2)))
    for i in range (0,r,2):
        for j in range (0,c,2):
            S[int(i/2),int(j/2)]=C[i:i+2,j:j+2].max()
        
    return S
```

```python
I=np.random.randint(1,5,(32,32)) #padding for 3x3 kernel
K=np.ones((5,5))
p=int(K.shape[0]/2)
C=f_conv2d(I,K,p)
b=np.random.randint(0,5)
C=C+b
C= f_Relu(C)
S=f_pool(C)
```
